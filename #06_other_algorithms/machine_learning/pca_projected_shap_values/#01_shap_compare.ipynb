```python
import numpy as np

def pca_time_scores_from_spatial(
    data_hw_t: np.ndarray,          # (H*W, T)
    spatial_h_w_k: np.ndarray,       # (H, W, K) 
    mean_hw: np.ndarray | None = None,   # (H*W,)
    std_hw: np.ndarray | None = None,    # (H*W,)
    whiten: bool = False,
    explained_var: np.ndarray | None = None,  # (K,) 
) -> np.ndarray:

    data_hw_t = np.asarray(data_hw_t)
    spatial_h_w_k = np.asarray(spatial_h_w_k)

    HW, T = data_hw_t.shape
    H, W, K = spatial_h_w_k.shape
    if H * W != HW:
        raise ValueError(f"shape mismatch: data has HW={HW}, spatial has H*W={H*W}")

    # X: (T, HW)
    X = data_hw_t.T.astype(np.float64, copy=False)

    if mean_hw is not None:
        mean_hw = np.asarray(mean_hw).reshape(1, HW)
        X = X - mean_hw
    if std_hw is not None:
        std_hw = np.asarray(std_hw).reshape(1, HW)
        X = X / std_hw

    Wmat = spatial_h_w_k.reshape(HW, K).astype(np.float64, copy=False)

    scores = X @ Wmat

    if whiten:
        if explained_var is None:
            raise ValueError("whiten=True requires explained_var (shape (K,))")
        scores = scores / np.sqrt(np.asarray(explained_var).reshape(1, K))

    return scores
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats  # ★これが必要

start_f, end_f = 0, 10

# 例: data と data2 は (n_sets, n_types, n) で用意されている前提
data = np.stack((np.mean(angle_10d_i[:, start_f:end_f], axis=1).transpose(), np.mean(odd_10d_i[:, start_f:end_f], axis=1).transpose(), np.mean(somato_10d_i[:, start_f:end_f], axis=1).transpose(), np.mean(SVT_rest[:, int(start_f*3/2):int(end_f*3/2)], axis=1).transpose()), axis=1)
data_null = np.stack((np.mean(angle_10d_null[:, start_f:end_f], axis=1).transpose(), np.mean(odd_10d_null[:, start_f:end_f], axis=1).transpose(), np.mean(somato_10d_null[:, start_f:end_f], axis=1).transpose(), np.mean(SVT_rest_n[:, int(start_f*3/2):int(end_f*3/2)], axis=1).transpose()), axis=1)

n_sets, n_types, n = 10, 4, 5

# ===== 見た目（あなたの設定を流用） =====
type_labels = ["Single", "Oddball", "Somato", "Spontaneous"]
bar_colors  = ["crimson", "steelblue", "orange", "black"]
sem_color   = ["crimson", "steelblue", "orange", "black"]
dot_color   = ["crimson", "steelblue", "orange", "black"]
dot_alpha   = 0.5
dot_size    = 15

figsize         = (12, 4)
bar_width       = 0.22
intra_gap       = 0.04
inter_group_gap = 0.45
cap_size        = 4
sig_line_color  = "black"
sig_fontsize    = 10

def p_to_stars(p):
    if p <= 0.005:
        return "***"
    elif p <= 0.01:
        return "**"
    elif p <= 0.05:
        return "*"
    else:
        return "n.s."

# ===== X位置（n_types本に合わせて作る） =====
group_centers = []
x_positions   = []
cursor = 0.0
for s in range(n_sets):
    left = cursor
    xs = [left + t*(bar_width + intra_gap) for t in range(n_types)]  # ★3本
    x_positions.append(xs)

    right_edge = left + (n_types-1)*(bar_width + intra_gap)          # ★3本ぶん
    cursor = right_edge + inter_group_gap
    group_centers.append(np.mean(xs))

x_positions   = np.asarray(x_positions)     # (n_sets, n_types)
group_centers = np.asarray(group_centers)

# ===== 集計 =====
means_all = data.mean(axis=2)                                # (n_sets, n_types)
sems_all  = data.std(axis=2, ddof=1) / np.sqrt(data.shape[2])

# 注記用の余白見積り（SEM込みで下限/上限を見積もる）
ymin = float((means_all - sems_all).min())
ymax = float((means_all + sems_all).max())
margin = (ymax - ymin) * 0.5 if ymax > ymin else 1.0

fig, ax = plt.subplots(figsize=figsize)

# ===== バー + SEM + 点 =====
for s in range(n_sets):
    xs = x_positions[s]
    for t in range(n_types):
        ax.bar(
            xs[t], means_all[s, t],
            width=bar_width, color="white",
            edgecolor=bar_colors[t], linewidth=0.6,
            yerr=sems_all[s, t], ecolor=sem_color[t], capsize=cap_size
        )
        vals = data[s, t]
        rng = np.random.default_rng(42 + s*17 + t)
        jitter = (rng.random(len(vals)) - 0.5) * (bar_width * 0.6)
        ax.scatter(
            np.full(len(vals), xs[t]) + jitter,
            vals, s=dot_size, alpha=dot_alpha, color=dot_color[t], zorder=3
        )

# ===== ★ ここが「data vs data2 の paired t-test を星で表示」 =====
# 星のy位置は「そのバーの平均+SEM」より少し上
star_offset = margin * 0.2  # 星を棒から離す距離（好みで調整）

for s in range(n_sets):
    xs = x_positions[s]
    for t in range(n_types):
        t_stat, p_val = stats.ttest_rel(data[s, t], data_null[s, t])  # paired t-test
        if np.isnan(p_val):
            p_val = 1.0

        label = p_to_stars(p_val)

        mean = means_all[s, t]
        sem  = sems_all[s, t]

        if mean >= 0:
            y  = mean + sem + star_offset
            va = "bottom"
        else:
            y  = mean - sem - star_offset
            va = "top"

        ax.text(xs[t], y, label, ha="center", va=va,
                color=sig_line_color, fontsize=sig_fontsize)

# ===== 軸など（あなたの設定を流用） =====
ax.set_ylim(ymin - margin*0.5, ymax + margin*1)
ax.set_xlim(x_positions.min() - bar_width, x_positions.max() + bar_width)

ax.set_xticks(group_centers)
ax.set_xticklabels([f"PC{i+1}" for i in range(n_sets)], rotation=0, ha="right", fontsize=11)
plt.setp(ax.get_xticklabels(), ha="center")
ax.set_xlabel("PC#", fontsize=14)
ax.set_ylabel("Normalized SHAP Values (a.u.)", fontsize=14)
ax.tick_params(axis="y", labelsize=10)

from matplotlib.patches import Patch
legend_handles = [Patch(facecolor=bar_colors[i], edgecolor=bar_colors[i], label=type_labels[i])
                  for i in range(n_types)]
ax.legend(handles=legend_handles, frameon=False, fontsize=12, title_fontsize=12, ncol=4, loc="upper left")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats  

start_f, end_f = 0, 10

data = np.stack((np.mean(d_angle_10d[:, start_f:end_f], axis=1).transpose(), np.mean(d_odd_10d[:, start_f:end_f], axis=1).transpose(), np.mean(d_somato_10d[:, start_f:end_f], axis=1).transpose(), np.mean(d_rest[:, int(start_f*3/2):int(end_f*3/2)], axis=1).transpose()), axis=1)
data_null = np.stack((np.mean(d_angle_n[:, start_f:end_f], axis=1).transpose(), np.mean(d_odd_n[:, start_f:end_f], axis=1).transpose(), np.mean(d_somato_n[:, start_f:end_f], axis=1).transpose(), np.mean(d_rest_n[:, int(start_f*3/2):int(end_f*3/2)], axis=1).transpose()), axis=1)

n_sets, n_types, n = 10, 4, 5

type_labels = ["Single", "Oddball", "Somato", "Spontaneous"]
bar_colors  = ["crimson", "steelblue", "orange", "black"]
sem_color   = ["crimson", "steelblue", "orange", "black"]
dot_color   = ["crimson", "steelblue", "orange", "black"]
dot_alpha   = 0.5
dot_size    = 15

figsize         = (12, 4)
bar_width       = 0.22
intra_gap       = 0.04
inter_group_gap = 0.45
cap_size        = 4
sig_line_color  = "black"
sig_fontsize    = 10

def p_to_stars(p):
    if p <= 0.005:
        return "***"
    elif p <= 0.01:
        return "**"
    elif p <= 0.05:
        return "*"
    else:
        return "n.s."

group_centers = []
x_positions   = []
cursor = 0.0
for s in range(n_sets):
    left = cursor
    xs = [left + t*(bar_width + intra_gap) for t in range(n_types)]
    x_positions.append(xs)

    right_edge = left + (n_types-1)*(bar_width + intra_gap)       
    cursor = right_edge + inter_group_gap
    group_centers.append(np.mean(xs))

x_positions   = np.asarray(x_positions)     # (n_sets, n_types)
group_centers = np.asarray(group_centers)

means_all = data.mean(axis=2)                                # (n_sets, n_types)
sems_all  = data.std(axis=2, ddof=1) / np.sqrt(data.shape[2])

ymin = float((means_all - sems_all).min())
ymax = float((means_all + sems_all).max())
margin = (ymax - ymin) * 0.5 if ymax > ymin else 1.0

fig, ax = plt.subplots(figsize=figsize)

for s in range(n_sets):
    xs = x_positions[s]
    for t in range(n_types):
        ax.bar(
            xs[t], means_all[s, t],
            width=bar_width, color="white",
            edgecolor=bar_colors[t], linewidth=0.6,
            yerr=sems_all[s, t], ecolor=sem_color[t], capsize=cap_size
        )
        vals = data[s, t]
        rng = np.random.default_rng(42 + s*17 + t)
        jitter = (rng.random(len(vals)) - 0.5) * (bar_width * 0.6)
        ax.scatter(
            np.full(len(vals), xs[t]) + jitter,
            vals, s=dot_size, alpha=dot_alpha, color=dot_color[t], zorder=3
        )

star_offset = margin * 0.2  

for s in range(n_sets):
    xs = x_positions[s]
    for t in range(n_types):
        t_stat, p_val = stats.ttest_rel(data[s, t], data_null[s, t])  # paired t-test
        if np.isnan(p_val):
            p_val = 1.0

        label = p_to_stars(p_val)

        mean = means_all[s, t]
        sem  = sems_all[s, t]

        if mean >= 0:
            y  = mean + sem + star_offset
            va = "bottom"
        else:
            y  = mean - sem - star_offset
            va = "top"

        ax.text(xs[t], y, label, ha="center", va=va,
                color=sig_line_color, fontsize=sig_fontsize)

ax.set_ylim(ymin - margin*0.5, ymax + margin*1)
ax.set_xlim(x_positions.min() - bar_width, x_positions.max() + bar_width)

ax.set_xticks(group_centers)
ax.set_xticklabels([f"PC{i+1}" for i in range(n_sets)], rotation=0, ha="right", fontsize=11)
plt.setp(ax.get_xticklabels(), ha="center")
ax.set_xlabel("PC#", fontsize=14)
ax.set_ylabel("Normalized SHAP Values (a.u.)", fontsize=14)
ax.tick_params(axis="y", labelsize=10)

from matplotlib.patches import Patch
legend_handles = [Patch(facecolor=bar_colors[i], edgecolor=bar_colors[i], label=type_labels[i])
                  for i in range(n_types)]
ax.legend(handles=legend_handles, frameon=False, fontsize=12, title_fontsize=12, ncol=4, loc="upper left")

ax.spines["top"].set_visible(False)
ax.spines["right"].set_visible(False)

plt.show()
```
